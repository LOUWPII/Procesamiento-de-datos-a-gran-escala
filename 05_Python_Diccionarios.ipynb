<h2 id="Dic">Diccionarios</h2>

<h3 id="content">¿Qué son los diccionarios?</h3>

Un diccionario esta conformado por llaves y valores. Un diccionario puede compararse a una lista. En vez de índices numéricos como los de las listas, los diccionarios tienen llaves. Estas llaves son usadas para acceder a los valores dentro de los diccionarios.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/DictsList.png" width="650" />

Un ejemplo de Diccionario <code>Dict</code>:

```python
# Crear el diccionario

Dict = {"key1": 1, "key2": "2", "key3": [3, 3, 3], "key4": (4, 4, 4), ('key5'): 5, (0, 1): 6}
Dict # Display el diccionario
```

Las llaves pueden ser cadenas:

```python
# Acceder a un valor mediante una llave

Dict["key1"] # Acceder al valor de 'key1'
```

Una llave puede ser cualquier tipo de objeto inmutable, pudiendo ser una tupla:

```python
# Acceder a un valor mediante una llave (tupla como llave)

Dict[(0, 1)] # Acceder al valor de la llave (0, 1)
```

Cada llave esta separada de su respectivo valor mediante dos puntos "<code>:</code>". Las comas separan los elementos y todo el diccionario esta encerrado usando llaves. Un diccionario vacío se representa escribiendo solo dos llaves, de esta forma "<code>{}</code>".

```python
# Crear un ejemplo de diccionario

release_year_dict = {"Thriller": "1982", "Back in Black": "1980", \
                    "The Dark Side of the Moon": "1973", "The Bodyguard": "1992", \
                    "Bat Out of Hell": "1977", "Their Greatest Hits (1971-1975)": "1976", \
                    "Saturday Night Fever": "1977", "Rumours": "1977"}
release_year_dict # Display el diccionario
```

Resumiendo, al igual que una lista, un diccionario guarda una secuencia de elementos. Cada uno es representado por una llave y su valor correspondiente. Los diccionarios se crean mediante dos llaves y estas contendrán las llaves y valores separados por una coma. A cada llave solo puede corresponderle un valor, sin embargo, múltiples llaves pueden tener el mismo. Las llaves solo pueden ser cadenas, números o tuplas, pero los valores pueden ser de cualquier tipo.

Es de ayuda visualizar un diccionario como una tabla, como se muestra en la imagen. La primera columna representa las llaves, la segunda los valores.

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/DictsStructure.png" width="650" />

<h3 id="key">Llaves</h3>

Los valores se pueden obtener mediante nombres:

```python
# Obtener el valor de la llave 'Thriller'

release_year_dict['Thriller']
```

Esto corresponde a:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/DictsKeyOne.png" width="500" />

De igual forma <b>The Bodyguard</b>

```python
# Obtener el valor mediante la llave 'The Bodyguard'

release_year_dict['The Bodyguard']
```

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/DictsKeyTwo.png" width="500" />

Ahora vamos a recuperar las llaves del diccionario mediante el método <code>release_year_dict()</code>:

```python
# Obtener todas las llaves del diccionario

release_year_dict.keys() # Retorna las llaves
```

Se puede obtener los valores mediante el método <code>values()</code>:

```python
# Obtener todos los valores del diccionario

release_year_dict.values() # Retorna los valores
```

Podemos agregar una entrada:

```python
# Agregar un valor con su llave en el diccionario

release_year_dict['Graduation'] = '2007' # Agrega 'Graduation': '2007'
release_year_dict # Display el diccionario actualizado
```

Tambien podemos borrar entradas:

```python
# Borrar entradas mediante la llave

del(release_year_dict['Thriller']) # Borra 'Thriller'
del(release_year_dict['Graduation']) # Borra 'Graduation'
release_year_dict # Display el diccionario actualizado
```

Podemos verificar si existe un elemento en el diccionario:

```python
# Verificar si existe la llave en el diccionario

'The Bodyguard' in release_year_dict # Verifica si 'The Bodyguard' es una llave
```

## Conclusiones sobre Diccionarios

Los diccionarios en Python son estructuras de datos muy potentes y flexibles, fundamentales para organizar información de manera eficiente.

### Casos de Uso:

*   **Representación de datos estructurados**: Ideales para almacenar registros o configuraciones donde cada pieza de información tiene un identificador único (llave), como perfiles de usuario, ajustes de aplicación o datos de un objeto.
*   **Búsqueda rápida**: Permiten un acceso extremadamente rápido a los valores mediante sus llaves, lo que los hace perfectos para escenarios donde la velocidad de consulta es crítica.
*   **Conteo de frecuencias**: Se utilizan comúnmente para contar la ocurrencia de elementos en una lista o cadena, usando los elementos como llaves y sus conteos como valores.
*   **Mapeo de datos**: Útiles para establecer relaciones uno a uno entre dos conjuntos de datos, como traducir códigos a descripciones o nombres a IDs.

### Complejidades Algorítmicas:

Los diccionarios en Python están implementados como tablas hash, lo que les confiere excelentes complejidades de tiempo en la mayoría de las operaciones:

*   **Acceso, inserción y eliminación**: En promedio, estas operaciones tienen una complejidad de tiempo de O(1). Esto significa que el tiempo que tardan no aumenta significativamente con el tamaño del diccionario.

*   **Peor caso**: En el peor de los casos, estas operaciones pueden degenerar a O(n), donde 'n' es el número de elementos en el diccionario. Esto ocurre si hay muchas colisiones en la tabla hash.
