<h3 id="content">Contenido de un Conjunto</h3>

Un conjunto es una colección de objetos únicos en Python. Un conjunto se define usando llaves de apertura y cierre <b>{}</b>. Python removerá automáticamente elementos duplicados.

```python
# Crear un conjunto

# Los conjuntos son colecciones de elementos únicos, ¡así que los duplicados se eliminan automáticamente. Es bueno para evitar repeticiones.
set1 = {"pop", "rock", "soul", "hard rock", "rock", "R&B", "rock", "disco"}
set1
```

El proceso de mapeado se muestra en la siguiente figura:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsUnique.png" width="1100" />

También se puede crear un conjunto a partir de una lista:

```python
# Convertir una lista en un conjunto

# Podemos tomar una lista (que sí permite duplicados y mantiene el orden) y convertirla en un conjunto.
# Al hacer esto, todos los elementos se vuelven únicos y se pierde el orden original. Útil para limpieza de datos.
album_list = [ "Michael Jackson", "Thriller", 1982, "00:42:19", \
              "Pop, Rock, R&B", 46.0, 65, "30-Nov-82", None, 10.0]
album_set = set(album_list)
album_set
```

Vamos a crear un conjunto de géneros musicales:

```python
# Convertir una lista de géneros musicales en un conjunto

# Otro ejemplo claro: aunque 'pop' y 'rock' están repetidos en la lista, en el conjunto solo aparecen una vez.
music_genres = set(["pop", "pop", "rock", "folk rock", "hard rock", "soul", \
                    "progressive rock", "soft rock", "R&B", "disco"])
music_genres
```

<h3 id="op">Operaciones con Conjuntos</h3>

Vamos a revisar las operaciones con conjuntos que nos ayudaran a modificarlo. Considera el conjunto <b>A</b>:

```python
# Conjunto de ejemplo

# Aquí creamos un conjunto inicial para empezar a jugar con las operaciones.
A = set(["Thriller", "Back in Black", "AC/DC"])
A
```

Podemos agregar un elemento al conjunto mediante el método <code>add()</code>:

```python
# Agregar un elemento al conjunto

# El método `add()` hace lo siguiente: si el elemento no está, lo añade. Si ya está, simplemente lo ignora y no da error
A.add("NSYNC")
A
```

Si agregamos el mismo elemento dos veces nada cambiara debido a que no puede haber elemento duplicados en un conjunto:

```python
# Trata de agregar un elemento duplicado al conjunto

# Intentamos añadir "NSYNC" de nuevo, pero como ya está, el conjunto no cambia. ¡Es la prueba de que mantienen solo elementos únicos!
A.add("NSYNC")
A
```

Podemos borrar un elemento del conjunto mediante el método <code>remove</code>:

```python
# Remover un elemento del conjunto

# Con `remove()` podemos quitar un elemento específico. Si se intenta remover algo que no existe, dará un error.
A.remove("NSYNC")
A
```

Podemos verificar si existe un elemento especifico dentro del conjunto con el comando <code>in</code>:

```python
# Verificar si existe un elemento en el conjunto

# El operador `in` es súper útil para saber rápidamente si un elemento ya forma parte de nuestro conjunto. Devuelve `True` o `False`.
"AC/DC" in A
```

<h3 id="logic">Operacion Lógicas con Conjuntos</h3>

Recuerda que puedes revisar la diferencia entre dos conjuntos, así como la diferencia simétrica, intersección y unión:

Considera estos dos conjuntos:

```python
# Conjuntos de ejemplo para operaciones lógicas

# Creamos dos conjuntos para explorar cómo interactúan entre sí.
album_set1 = set(["Thriller", 'AC/DC', 'Back in Black'])
album_set2 = set([ "AC/DC", "Back in Black", "The Dark Side of the Moon"])
```

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsSamples.png" width="650" />

```python
# Imprimir dos conjuntos

# Simplemente mostramos ambos conjuntos para ver con qué estamos trabajando antes de las operaciones.
album_set1, album_set2
```

Como los dos conjuntos contienen <b>AC/DC</b> y <b>Back in Black</b> podemos representar estos elementos en común en la intersección de dos círculos.

<img src = "https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsLogic.png" width = "650" />

Para encontrar la intersección de dos conjuntos usaremos <code>&</code>:

```python
# Encontrar la intersección

# La intersección nos da los elementos que están presentes en ambos conjuntos.
intersection = album_set1 & album_set2
intersection
```

Puedes encontrar los elementos que únicamente están dentro de <code>album_set1</code> usando el método <code>difference</code>:

```python
# Encuentra la diferencia en set1 pero no en set2

# `difference()` nos muestra los elementos que están en el primer conjunto, pero no en el segundo.
album_set1.difference(album_set2)
```

Solo tienes que considerar los elementos en <code>album_set1</code>; todos los de <code>album_set2</code>, incluyendo los de la intersección, no están incluidos.

Los elementos que pertenecen a <code>album_set2</code> pero no en <code>album_set1</code> esta dado por:

```python
album_set2.difference(album_set1)
# Y aquí, al revés: elementos exclusivos del segundo conjunt que no están en el primero
```

<img src = "https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsRight.png" width="650" />

Puedes encontrar la intersección de <code>album_list1</code> y <code>album_list2</code> con el método <code>intersection</code>:

```python
# Usar el método intersection() para encontrar la intersección entre de album_set1 y album_set2

# Este es otro modo de obtener la intersección, el método `.intersection()` hace lo mismo que el operador `&`.
album_set1.intersection(album_set2)
```

Lo siguiente corresponde a la intersección de los dos círculos:

<img src="https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsIntersect.png" width="650" />

La unión es la suma de todos los elementos de ambos conjuntos, se representa pintando los dos círculos:

<img src = "https://s3-api.us-geo.objectstorage.softlayer.net/cf-courses-data/CognitiveClass/PY0101EN/Chapter%202/Images/SetsUnion.png" width="650" />

La unión esta dada por:

```python
# Encontrar la unión de los dos conjuntos

# `union()` combina todos los elementos únicos de ambos conjuntos.
album_set1.union(album_set2)
```

Se puede averiguar si un conjunto es un súper conjunto o un sub conjunto de otro, respectivamente, de la siguiente forma:

```python
# Revisar si es un superconjunto

# `issuperset()` nos dice si un conjunto contiene todos los elementos de otro.
set(album_set1).issuperset(album_set2)
```

```python
# Revisar si es un subconjunto

# `issubset()` es lo contrario
set(album_set2).issubset(album_set1)
```

Este es un ejemplo donde <code>issubset()</code> y <code>issuperset()</code> regresan un valor verdadero:

```python
# Revisar si es un subconjunto (ejemplo True)

# Aquí vemos un ejemplo donde sí es un subconjunto.
set({"Back in Black", "AC/DC"}).issubset(album_set1)
```

```python
# Revisar si es un superconjunto (ejemplo True)

# Y aquí un ejemplo de superconjunto. `album_set1` contiene a los dos elementos del conjunto más pequeño.
album_set1.issuperset({"Back in Black", "AC/DC"})
```

Considera que la lista <code>A = [1, 2, 2, 1]</code> y el conjunto <code>B = set([1, 2, 2, 1])</code> suman <code>sum(A) = sum(B)</code>

```python
# Escribe abajo tu código y presiona Shift+Enter para ejecutarlo

# Analicemos si `sum(A)` es igual a `sum(B)`.
A = [1, 2, 2, 1]  # Una lista, permite duplicados
B = set([1, 2, 2, 1]) # Un conjunto, los duplicados se eliminan automáticamente (será {1, 2})

# Sumamos la lista. Aquí 1+2+2+1 = 6
sum_A = sum(A)
# Sumamos el conjunto. Aquí 1+2 = 3.
sum_B = sum(B)

print(f"La suma de A es: {sum_A}")
print(f"La suma de B es: {sum_B}")
print(f"¿sum(A) == sum(B)?: {sum_A == sum_B}")
```

Los sets tienen varios casos de uso, entre estos:

- Limpieza de datos: Para eliminar elementos duplicados de listas o bases de datos, asegurando que cada valor sea único.

- Análisis de datos: Para encontrar elementos comunes (intersección), elementos únicos (diferencia) o combinar colecciones (unión) de manera eficiente.

- Verificación de pertenencia: Comprobar rápidamente si un elemento existe en una colección grande, lo cual es muy eficiente.

- Generación de IDs únicos: Cuando se necesita una colección de identificadores que no deben repetirse.

- Rendimiento: Los conjuntos están implementados como tablas hash, lo que les confiere un rendimiento excelente para las operaciones básicas:

- Inserción (add()): Muy rápido, en promedio O(1).
- Eliminación (remove()): Muy rápido, en promedio O(1).
- Verificación de pertenencia (in): Muy rápido, en promedio O(1).
